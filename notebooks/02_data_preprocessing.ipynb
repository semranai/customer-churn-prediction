```python
import pandas as pd
import numpy as np

from pathlib import Path
from sklearn.model_selection import train_test_split

project_root = Path.cwd().parent

file_path = (
    project_root
    / "data"
    / "raw"
    / "WA_Fn-UseC_-Telco-Customer-Churn.csv"
)

df = pd.read_csv(file_path)

df_clean = df.copy()

df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df_clean.shape)
```

```python
X = df_clean.drop(columns=["customerID", "Churn"])

y = df_clean["Churn"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
numerical_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

print(numerical_transformer)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

print(numerical_transformer)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

print(categorical_transformer)
```

```python
print("Numerical Pipeline:")
print(numerical_transformer)

print("\nCategorical Pipeline:")
print(categorical_transformer)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

print(categorical_transformer)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print(preprocessor)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

print("Original training shape:", X_train.shape)

print("Processed training shape:", X_train_processed.shape)
```

```python
X_test_processed = preprocessor.transform(X_test)

print("Original testing shape:", X_test.shape)

print("Processed testing shape:", X_test_processed.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Total processed features:", len(feature_names))

print("\nProcessed Feature Names:")

for feature in feature_names:
    print(feature)
```

```python
print("Training data:")
print("Before:", X_train.shape)
print("After:", X_train_processed.shape)

print("\nTesting data:")
print("Before:", X_test.shape)
print("After:", X_test_processed.shape)

print("\nTotal features:", len(feature_names))
```

## 7. Machine Learning Model — Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print(logistic_model)
```

```python
logistic_model.fit(
    X_train_processed,
    y_train
)

print("Model trained successfully!")
```

```python
y_pred = logistic_model.predict(
    X_test_processed
)

print("First 10 predictions:")

print(y_pred[:10])
```

```python
logistic_model.predict(X_test_processed)
```

```python
import pandas as pd

results = pd.DataFrame({
    "Actual_Churn": y_test.to_numpy(),
    "Predicted_Churn": y_pred
})

results.head(10)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Model Accuracy: {accuracy:.2%}")
```

```python
correct_predictions = (
    results["Actual_Churn"]
    == results["Predicted_Churn"]
).sum()

incorrect_predictions = (
    results["Actual_Churn"]
    != results["Predicted_Churn"]
).sum()

print("Correct Predictions:", correct_predictions)

print("Incorrect Predictions:", incorrect_predictions)

print("Total Predictions:", len(results))
```

```python
print("MODEL SUMMARY")
print("----------------------------")

print("Algorithm: Logistic Regression")

print("Training customers:", len(X_train))

print("Testing customers:", len(X_test))

print("Processed features:", X_train_processed.shape[1])

print(f"Accuracy: {accuracy:.2%}")

print("Correct predictions:", correct_predictions)

print("Incorrect predictions:", incorrect_predictions)
```
